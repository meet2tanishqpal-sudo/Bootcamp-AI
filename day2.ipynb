```python
import numpy as np

# Create a dataset
players = np.array(['Messi', 'Ronaldo', 'Neymar', 'Mbappe', 'Lewandowski'])

matches = np.array([1000, 1200, 800, 600, 700])

np.random.seed(19)  # For reproducibility

num = np.random.randint(10, 100, size=len(players))

num1 = np.random.randint(10, 100, size=len(matches))

print(num)  
print(num1)
```

```python
data = np.array([num1, num])

print("Dataset:")
print(data)

# Row-wise
print("Row-wise Total:", np.sum(data, axis=1))
print("Row-wise Average:", np.mean(data, axis=1))
print("Row-wise Standard Deviation:", np.std(data, axis=1))

# Column-wise
print("Column-wise Total:", np.sum(data, axis=0))
print("Column-wise Average:", np.mean(data, axis=0))
print("Column-wise Standard Deviation:", np.std(data, axis=0))
```

```python
import matplotlib.pyplot as plt
```

```python
time = np.linspace(0, 10, 500)

amplitude1 = np.sin(np.pi * time)

amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)
```

```python
#create the canvas
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(time, amplitude1, color='crimson', linewidth=2, label="Sensor A Displacement")
ax.plot(time, amplitude2, color='blue', linewidth=2, label="Sensor B Displacement")
ax.set_title('Damped Harmonic Response of Structure', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (mm)', fontsize=12)
# Enable engineering gridlines
ax.grid(True, linestyle='--', alpha=0.6)
# Configure the Legend
ax.legend(loc='upper right', frameon=True, shadow=True)
```
