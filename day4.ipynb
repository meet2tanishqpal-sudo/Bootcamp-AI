```python
import cv2

# Load the image
img = cv2.imread(
    r'C:\Users\User\OneDrive\Desktop\Bootcamp AI\imagee.jpeg'
)

# Load OpenCV's pre-trained face detector
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30, 30)
)

# Draw rectangle around each detected face
for (x, y, w, h) in faces:
    cv2.rectangle(
        img,
        (x, y),
        (x + w, y + h),
        (255, 0, 0),
        2
    )

# Display result
cv2.imshow("Face Detection", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

face = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

cam = cv2.VideoCapture(0)

cv2.namedWindow("Live Face Detection", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Live Face Detection", 640, 480)

while True:
    ret, frame = cam.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face.detectMultiScale(gray, 1.1, 5)

    for x, y, w, h in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 2)

    cv2.imshow("Live Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cam.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2

model = YOLO("yolo11n.pt")

cap = cv2.VideoCapture(0)

cv2.namedWindow("Object Detection", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Object Detection", 640, 480)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame)
    frame = results[0].plot()

    cv2.imshow("Object Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp

# Hand detector
options = mp.tasks.vision.HandLandmarkerOptions(
    base_options=mp.tasks.BaseOptions(
        model_asset_path="hand_landmarker.task"
    ),
    num_hands=1
)

detector = mp.tasks.vision.HandLandmarker.create_from_options(options)

# Camera
cap = cv2.VideoCapture(0)

canvas = None
px, py = 0, 0

cv2.namedWindow("Air Canvas", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Air Canvas", 640, 480)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame = cv2.flip(frame, 1)

    if canvas is None:
        canvas = frame.copy()
        canvas[:] = 0

    # Convert frame to MediaPipe image
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    # Detect hand
    result = detector.detect(image)

    if result.hand_landmarks:
        # Index finger tip = landmark 8
        tip = result.hand_landmarks[0][8]

        h, w, _ = frame.shape
        x = int(tip.x * w)
        y = int(tip.y * h)

        if px != 0:
            cv2.line(canvas, (px, py), (x, y), (255, 0, 255), 5)

        px, py = x, y

        cv2.circle(frame, (x, y), 8, (0, 255, 0), -1)

    else:
        px, py = 0, 0

    # Combine camera and drawing
    output = cv2.add(frame, canvas)

    cv2.imshow("Air Canvas", output)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
detector.close()
cv2.destroyAllWindows()
```

```python
# Install if needed:
# !pip install nltk wordcloud matplotlib

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Download NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Your academic and personal journey
text = """
My academic journey started at DPS Indirapuram, where I developed a strong interest
in technology, learning, creativity, and communication. I am currently pursuing
BTech in Computer Science and Engineering at SRM Institute of Science and Technology
Ghaziabad.

During my college journey, I have explored Python, C++, Data Science, Machine Learning,
Artificial Intelligence, Web Development, Cybersecurity, and Software Development.

My journey also includes public speaking and leadership through Toastmasters, where I
have taken roles such as PR member, Timer, Table Topics Master, and Grammarian.

I have participated in hackathons, technical competitions, and technology events.
I worked on projects such as PustakSetu, an AI powered book donation platform, and
an AI powered road accident detection and emergency response system.

I have also explored robotics and worked on ideas related to autonomous restaurant
robots and AI based solutions.

My journey with Google AI and Gemini has given me opportunities to learn, build,
share, and inspire others. Becoming a Google Student Ambassador has been an important
part of my technology journey.

My interests include Artificial Intelligence, Machine Learning, Data Science,
Web Development, Robotics, Cybersecurity, and problem solving.

I continue to improve my programming skills, practice Data Structures and Algorithms,
participate in competitions, build projects, and explore new technologies.

My long term goal is to become a strong software engineer, gain industry experience
through internships, build innovative products, and eventually create my own startup.
"""

# Convert text to lowercase
text = text.lower()

# Tokenize text into individual words
words = word_tokenize(text)

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords and non-alphabetic words
filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

# Count frequency of each word
word_freq = Counter(filtered_words)

print("Word Frequencies:")
print(word_freq)

# Create Word Cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    max_words=100
).generate_from_frequencies(word_freq)

# Display Word Cloud
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
