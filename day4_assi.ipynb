```python

import cv2

# Load image
img = cv2.imread(
    r'C:\Users\User\OneDrive\Desktop\Bootcamp AI\imagee.jpeg'
)

if img is None:
    print("Image could not be loaded.")
else:
    print("Image loaded successfully")
    print("Original shape:", img.shape)

    # 1. Resize
    resized = cv2.resize(img, (640, 480))

    # 2. Crop
    cropped = img[100:500, 200:800]

    # 3. Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 4. Blur
    blurred = cv2.GaussianBlur(img, (15, 15), 0)

    # 5. Rotate
    rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # 6. Increase brightness
    bright = cv2.convertScaleAbs(img, alpha=1.2, beta=40)

    # Display images
    cv2.imshow("Original", img)
    cv2.imshow("Resized", resized)
    cv2.imshow("Cropped", cropped)
    cv2.imshow("Grayscale", gray)
    cv2.imshow("Blurred", blurred)
    cv2.imshow("Rotated", rotated)
    cv2.imshow("Brightened", bright)

    # Save edited images
    cv2.imwrite("resized.jpg", resized)
    cv2.imwrite("gray.jpg", gray)
    cv2.imwrite("blurred.jpg", blurred)
    cv2.imwrite("rotated.jpg", rotated)
    cv2.imwrite("bright.jpg", bright)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
